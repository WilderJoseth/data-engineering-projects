# Bronze setup

```python
%%sql
CREATE SCHEMA IF NOT EXISTS bronze;
```

```python
%%sql
CREATE TABLE IF NOT EXISTS bronze.metadata_objects
(
    source_database      STRING NOT NULL,
    source_schema        STRING NOT NULL,
    object_name          STRING NOT NULL,
    object_type          STRING NOT NULL,
    source_object_id     INT,
    is_system_object     BOOLEAN NOT NULL,
    source_modified_at   TIMESTAMP,
    execution_step_id    BIGINT NOT NULL,
    created_at           TIMESTAMP NOT NULL,
    created_by           STRING
)
USING DELTA;
```

```python
%%sql
CREATE TABLE IF NOT EXISTS bronze.metadata_columns
(
    source_database      STRING NOT NULL,
    source_schema        STRING NOT NULL,
    object_name          STRING NOT NULL,
    object_type          STRING NOT NULL,
    source_object_id     INT,
    column_name          STRING NOT NULL,
    ordinal_position     INT NOT NULL,
    data_type            STRING NOT NULL,
    max_length           INT,
    precision            INT,
    scale                INT,
    is_nullable          BOOLEAN,
    is_identity          BOOLEAN,
    is_computed          BOOLEAN,
    default_definition   STRING,
    computed_definition  STRING,
    execution_step_id    BIGINT NOT NULL,
    created_at           TIMESTAMP NOT NULL,
    created_by           STRING
)
USING DELTA;
```

```python
%%sql
CREATE TABLE IF NOT EXISTS bronze.metadata_constraints
(
    source_database           STRING NOT NULL,
    source_schema             STRING NOT NULL,
    object_name               STRING NOT NULL,
    source_object_id          INT,
    constraint_name           STRING NOT NULL,
    constraint_type           STRING NOT NULL,
    referenced_database       STRING,
    referenced_schema         STRING,
    referenced_object         STRING,
    referenced_object_id      INT,
    delete_action             STRING,
    update_action             STRING,
    execution_step_id         BIGINT NOT NULL,
    created_at                TIMESTAMP NOT NULL,
    created_by                STRING
)
USING DELTA;
```

```python
%%sql
CREATE TABLE IF NOT EXISTS bronze.metadata_constraint_columns
(
    source_database           STRING NOT NULL,
    source_schema             STRING NOT NULL,
    object_name               STRING NOT NULL,
    constraint_name           STRING NOT NULL,
    column_name               STRING NOT NULL,
    referenced_column_name    STRING,
    ordinal_position          INT NOT NULL,
    execution_step_id         BIGINT NOT NULL,
    created_at                TIMESTAMP NOT NULL,
    created_by                STRING
)
USING DELTA;
```

```python
%%sql
CREATE TABLE IF NOT EXISTS bronze.metadata_relationships
(
    source_database       STRING NOT NULL,
    source_schema         STRING NOT NULL,
    source_object         STRING NOT NULL,
    source_column         STRING,
    target_database       STRING,
    target_schema         STRING,
    target_object         STRING NOT NULL,
    target_column         STRING,
    relationship_type     STRING NOT NULL,
    relationship_name     STRING,
    execution_step_id     BIGINT NOT NULL,
    created_at            TIMESTAMP NOT NULL,
    created_by            STRING
)
USING DELTA;
```

```python
%%sql
CREATE TABLE IF NOT EXISTS bronze.metadata_modules
(
    source_database       STRING NOT NULL,
    source_schema         STRING NOT NULL,
    object_name           STRING NOT NULL,
    object_type           STRING NOT NULL,
    source_object_id      INT,
    module_definition     STRING NOT NULL,
    execution_step_id     BIGINT NOT NULL,
    created_at            TIMESTAMP NOT NULL,
    created_by            STRING
)
USING DELTA;
```

```python
%%sql
CREATE TABLE IF NOT EXISTS bronze.content_files
(
    source_system       STRING NOT NULL,
    source_container    STRING NOT NULL,
    source_path         STRING NOT NULL,
    file_name           STRING NOT NULL,
    file_type           STRING,
    file_size_bytes     BIGINT,
    bronze_path         STRING NOT NULL,
    execution_step_id   BIGINT NOT NULL,
    created_at          TIMESTAMP NOT NULL,
    created_by          STRING
)
USING DELTA;
```
