Define reusable validation utilities for Fabric validation notebooks. This notebook contains imports and function definitions only; callers own execution state and result collections.

## Imports

```python
from pyspark.sql import functions as F
```

## Validation Helpers

Combine temporary `is_*_passed` flags and append validation summaries to a caller-owned list.

```python
def get_validation_condition(df):
    """Build a condition requiring every temporary validation flag to pass."""

    validation_flag_columns = [
        column_name
        for column_name in df.columns
        if column_name.startswith("is_")
        and column_name.endswith("_passed")
    ]

    condition = F.lit(True)

    for column_name in validation_flag_columns:
        condition = condition & F.col(column_name)

    return condition
```

```python
def add_validation_result(
    validation_results,
    details,
    affected_row_count,
    validation_code
):
    """Append a validation summary to validation_results.

    Mutate the supplied list and return None.
    """

    validation_results.append({
        "details": details,
        "affected_row_count": affected_row_count,
        "validation_code": validation_code
    })
```
