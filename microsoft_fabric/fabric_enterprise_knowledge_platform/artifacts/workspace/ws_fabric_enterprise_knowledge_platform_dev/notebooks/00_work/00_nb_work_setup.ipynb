```python
%%sql
CREATE SCHEMA IF NOT EXISTS work;
```

```python
%%sql
CREATE TABLE IF NOT EXISTS work.metadata_objects
(
    object_id                   STRING NOT NULL,
    source_database             STRING NOT NULL,
    source_schema               STRING NOT NULL,
    object_name                 STRING NOT NULL,
    object_type                 STRING NOT NULL,
    source_object_id            INT,
    is_system_object            BOOLEAN,
    source_modified_at          TIMESTAMP,
    source_execution_step_id    BIGINT NOT NULL
)
USING DELTA
PARTITIONED BY (source_database);
```

```python
%%sql
CREATE TABLE IF NOT EXISTS work.metadata_columns
(
    column_id                   STRING NOT NULL,
    object_id                   STRING NOT NULL,
    source_database             STRING NOT NULL,
    source_schema               STRING NOT NULL,
    object_name                 STRING NOT NULL,
    column_name                 STRING NOT NULL,
    ordinal_position            INT NOT NULL,
    data_type                   STRING NOT NULL,
    max_length                  INT,
    precision                   INT,
    scale                       INT,
    is_nullable                 BOOLEAN,
    is_identity                 BOOLEAN,
    is_computed                 BOOLEAN,
    default_definition          STRING,
    computed_definition         STRING,
    source_execution_step_id    BIGINT NOT NULL
)
USING DELTA
PARTITIONED BY (source_database);
```

```python
%%sql
CREATE TABLE IF NOT EXISTS work.metadata_constraints
(
    constraint_id               STRING NOT NULL,
    object_id                   STRING NOT NULL,
    source_database             STRING NOT NULL,
    source_schema               STRING NOT NULL,
    object_name                 STRING NOT NULL,
    constraint_name             STRING NOT NULL,
    constraint_type             STRING NOT NULL,
    referenced_database         STRING,
    referenced_schema           STRING,
    referenced_object           STRING,
    referenced_object_id        STRING,
    delete_action               STRING,
    update_action               STRING,
    source_execution_step_id    BIGINT NOT NULL
)
USING DELTA
PARTITIONED BY (source_database);
```

```python
%%sql
CREATE TABLE IF NOT EXISTS work.metadata_constraint_columns
(
    constraint_column_id        STRING NOT NULL,
    constraint_id               STRING NOT NULL,
    column_id                   STRING NOT NULL,
    source_database             STRING NOT NULL,
    source_schema               STRING NOT NULL,
    object_name                 STRING NOT NULL,
    constraint_name             STRING NOT NULL,
    column_name                 STRING NOT NULL,
    referenced_column_name      STRING,
    ordinal_position            INT NOT NULL,
    source_execution_step_id    BIGINT NOT NULL
)
USING DELTA
PARTITIONED BY (source_database);
```

```python
%%sql
CREATE TABLE IF NOT EXISTS work.metadata_relationships
(
    relationship_id             STRING NOT NULL,
    source_database             STRING NOT NULL,
    source_schema               STRING NOT NULL,
    source_object               STRING NOT NULL,
    source_column               STRING,
    target_database             STRING,
    target_schema               STRING,
    target_object               STRING NOT NULL,
    target_column               STRING,
    relationship_type           STRING NOT NULL,
    relationship_name           STRING,
    source_execution_step_id    BIGINT NOT NULL
)
USING DELTA
PARTITIONED BY (source_database);
```

```python
%%sql
CREATE TABLE IF NOT EXISTS work.metadata_modules
(
    module_id                   STRING NOT NULL,
    object_id                   STRING NOT NULL,
    source_database             STRING NOT NULL,
    source_schema               STRING NOT NULL,
    object_name                 STRING NOT NULL,
    object_type                 STRING NOT NULL,
    source_object_id            STRING,
    module_definition           STRING NOT NULL,
    source_execution_step_id    BIGINT NOT NULL
)
USING DELTA
PARTITIONED BY (source_database);
```

```python
%%sql
CREATE TABLE IF NOT EXISTS work.content_documents
(
    content_id                  STRING NOT NULL,
    source_container            STRING NOT NULL,
    source_path                 STRING NOT NULL,
    bronze_path                 STRING NOT NULL,
    content_path                STRING NOT NULL,
    file_name                   STRING NOT NULL,
    file_extension              STRING NOT NULL,
    content                     STRING,
    source_execution_step_id    BIGINT NOT NULL
)
USING DELTA;
```
